#### Simplisel Kompleks
Simplisel kompleks, basit yapıtaşlarından (nokta, çizgi, üçgen vb.) oluşan bir geometrik yapıdır
. Örneğin, bir 0-simplex tek bir noktayı; 1-simplex iki noktayı birleştiren doğru parçasını; 2-simplex ise üç noktayı birleştirerek oluşturulan bir üçgeni ifade eder
. Bu yapıtaşları bir araya gelerek daha büyük şekiller inşa eder. Aşağıda bazı temel basit yapıtaşları örneklendirilmiştir:

•0-simplex: Tek bir nokta (bağımsız bir konum).

•1-simplex: İki nokta arasındaki doğru parçası.

•2-simplex: Üç noktayı birleştirerek oluşturulan üçgen.

Bu örnekler, bir basitsel kompleksin noktalar birleştikçe çizgilere ve üçgenlere dönüştüğünü, adım adım büyüyen bir yapı olduğunu gösterir.





#### Filtrasyon
Filtrasyon, bir basitsel kompleksin zaman içinde veya bir eşik parametreye bağlı olarak kademeli büyüme sürecidir
. Örneğin bir nokta bulutunda başlangıçta sadece ayrı noktalar (0-simplex’ler) varken, eşik değeri arttıkça noktalar arası mesafe koşulu sağlanır ve yeni bağlantılar eklenir. Tipik bir filtrasyon adımları şunlardır:
1-Belirli bir mesafe eşiği R belirlenir

2-Aralarında mesafe R’den küçük olan her iki noktayı bir doğru parçası ile birleştirilir.

3-Üç noktanın birbirleriyle çiftler halinde bağlı olması durumunda aralarına bir üçgen (2-simplex) eklenir.

4-Bu işlemi daha yüksek boyutlu basit yapı taşları için tekrarlanır (örneğin tetrahedron eklenir).

Bu şekilde her aşamada eşik büyüdükçe kompleks yeni bağlantılar ve yüzeylerle genişler. Filtrasyon sayesinde, karmaşık bir yapının oluşumu adım adım takip edilebilir ve her adımda eklenen topolojik özellikler görülebilir.




#### Kalıcı Homoloji
Kalıcı homoloji, filtrasyon sürecinde ortaya çıkan topolojik özelliklerin (bağımsız bileşenler, delikler vb.) ne kadar süreyle var olduğunu ölçer. Her bir özellik için ortaya çıkış (doğum) ve kayboluş (ölüm) zamanları kaydedilir. Örneğin başlangıçta her nokta ayrı bir bileşen olarak başlar; eşik arttıkça bu bileşenler birleşir ve sayıları azalır. Bu süreçte her bileşenin veya deliğin ortaya çıktığı ve yok olduğu anda tutulan kayıt, söz konusu özelliğin “yaşını” (ömrünü) verir. Uzun ömürlü (persistent) özellikler, verinin gerçek yapısına ait kabul edilirken, kısa ömürlü olanlar genellikle gürültü olarak değerlendirilir. Kalıcı homoloji, bu özelliklerin doğum ve ölüm bilgilerini bir arada değerlendirerek verideki anlamlı topolojik yapıları ortaya çıkarmayı sağlar.





#### Betti Sayıları
Betti sayıları, bir uzayın belirli boyutlardaki topolojik boşluklarını sayısal olarak ifade eder. β₀, bağımsız bileşen (bağlantısız parça) sayısını; β₁, bir boyutlu delik (döngü/halka) sayısını; β₂ ise iki boyutlu boşluk sayısını gösterir. Örneğin:

•Üç ayrı nokta için β₀ = 3 (üç bağımsız bileşen),   β₁ = 0,   β₂ = 0.

•Tek bir halka (çember) için β₀ = 1 (tek parça),    β₁ = 1 (bir delik),    β₂ = 0.

•Kapalı bir küre yüzeyi için β₀ = 1,    β₁ = 0,     β₂ = 1 (küresel boşluk).

Bu sayılar, basitçe her boyuttaki “delik” sayısını ifade eder ve karmaşık şekillerin temel topolojik özelliklerini özetler.




#### Kalıcılık Diyagramı
Persistence diyagramı, filtrasyondaki topolojik özelliklerin doğum ve ölüm zamanlarını grafiksel olarak gösterir. Bir diyagramda her bir nokta bir topolojik özelliğe karşılık gelir; X ekseni özelliğin doğduğu (uygun eşik değerini bulduğu) zamanı, Y ekseni ise öldüğü (kaybolduğu) zamanı gösterir. Örneğin bir bağlı bileşen yeni bir bağlantıyla ortaya çıktığında diyagramda (doğum zamanı) işaretlenir; başka bir bileşenle birleşip ortadan kalktığında (ölüm zamanı) gösterilir. Diyagramdaki bir noktanın diyagonal (y=x) çizgiden uzaklığı, o özelliğin ne kadar uzun süre var olduğuna işaret eder: noktalar diyagondan ne kadar uzunsa, özellik o kadar uzun ömürlüdür ve veri için o kadar anlamlı kabul edilir. Kısa ömürlü özellikler diyagona yakın noktalar olarak görülürken, uzun ömürlü özellikler diyagramda diyagonun uzağında belirir. Böylece persistence diyagramı, önemli topolojik yapıları (uzun ömürlü delikler vb.) ayırt etmeye yarayan etkili bir araçtır.

# TEORİ

## Simplisel Kompleks


Bir şekli anlamak istiyorsak onu parçalarına ayırmak isteriz. Simplicial kompleks, bunu parçala ve yönet felsefesiyle yapar:    
Noktalarla başlar → nokta bir 0-simplekstir     
Noktaları birleştirip kenar oluşturursun → bir 1-simpleks   
Üç kenarla bir üçgen kapatırsın → bir 2-simpleks    
Dört üçgeni bir hacimsel yapı olacak şekilde birleştirirsin → bir 3-simpleks (dörtyüzlü, tetrahedron)

Bunları doğru kurallarla birleştirirsek, ortaya çıkan yapı bir simplicial kompleks olur.


•Simplicial kompleksler, karmaşık geometrik/topolojik yapıları:
Bilgisayarda temsil etmeyi kolaylaştırır
Hesaplanabilir hale getirir (örneğin homoloji, persistent homoloji)
Nokta bulutları gibi ham veriden topolojik çıkarımlar yapmamıza izin verir

•Özellikle topolojik veri analizi (TDA), simplicial kompleksler sayesinde:
Veri içindeki bağlantı, döngü, boşluk gibi yapıları sezebilir
Bunu cebirsel olarak analiz edebilir.

**Tanım:(k-boyutlu Simpleks):** $ℝ^n$ içinde $v_0$, $v_1$, ..., $v_k$ noktaları doğrusal olarak bağımsız (yani affine bağımsız) olmak üzere, bu noktaların konveks birleşimi olan küme:

$$
\sigma = \left\{ \sum_{i=0}^{k} t_i v_i \mid t_i \ge 0, \sum t_i = 1 \right\}
$$

bir k-boyutlu simpleks olarak adlandırılır ve [$v_0$, $v_1$, ..., $v_k$] ile gösterilir.

**Tanım:(Simplisel Kompleks):** Bir simplisel kompleks K, aşağıdaki iki koşulu da sağlayan bir simpleks ailesidir:

•Her simpleksin tüm *yüzleri(1)* yine bu ailede kalır.

•Her iki simpleksin birleşimi, ya boş kümedir ya da ikisinin ortak *bir yüzüdür(2)* .

*(1)Yüz:* Bir simpleksin tepe noktalarının herhangi bir alt kümesiyle oluşturulan alt-simpleks. Örneğin;    
[A, B, C] üçgeninin yüzleri: &nbsp;[A, B], [B, C], [C, A] (kenarlar)&nbsp;&nbsp; ve&nbsp;&nbsp; [A], [B], [C] (köşeler)

*(2)Ortak Yüz:* İki simpleksin kesişimi bir yüzse, bunlar düzgün birleşmiştir demektir.

**🔹Örnek:** Aşağıdaki yapı bir simplisel komplekstir:

→Noktalar: $v_0$, $v_1$, $v_2$

→Kenarlar: [$v_0$, $v_1$], [$v_1$, $v_2$], [$v_0$, $v_2$]

→Üçgen: [$v_0$, $v_1$, $v_2$]

•Her kenarın uç noktaları kompleks içinde ✅        
•Üçgenin kenarları kompleks içinde ✅       
•Her iki simpleksin kesişimi bir kenar ya da köşe ✅


**🔹Örnek:** İki üçgen yalnızca bir noktada kesişiyorsa simplisel kompleks değildir.

## Zincir Grupları

Bir simplisel komplekste elimizde birçok yapı var: noktalar (0-simpleks), kenarlar (1-simpleks), üçgenler(2-simpleks) vs.
Bu yapıları "topolojik veri" olarak görüp üzerinde işlem yapmak istiyorsak her boyuttaki simplekslerin cebirsel kombinasyonlarını alabileceğimiz gruplar kurarız. Bunlara zincir grubu denir.

**Tanım:(Serbest abel gurubu):** Her elemanı temel vektörse yani katsayıları tam sayı ise o gruba serbest abel grubu denir. Vektörlerin lineer kombinasyonlarındaki $c_1$, $c_2$ gibi katsayılar tam sayıdır.

**Tanım:** Bir simplisel kompleks K verilsin. K'deki tüm k-simplekslerin serbest abel gurubu olan $C_k$(K), K'in k-zincir grubudur.


**🔹Örnek:**    
Diyelim ki K simplisel kompleksinde şu kenarlar olsun; $σ_1$ = [$v_0$, $v_1$],&nbsp;&nbsp;&nbsp; $σ_2$=[$v_1$, $v_2$],&nbsp;&nbsp;&nbsp; $σ_3$=[$v_2$_, $v_0$].    
O zaman;    
$C_1$(K) = $α_1$.$σ_1$ + $α_2$.$σ_2$ + $α_3$.$σ_3$ ,     &nbsp;&nbsp;&nbsp;&nbsp;          a_i ∈ Z

**Tanım:(Zincir):** k-simplekslerin tamsayı katsayılarla alınmış toplamına zincir denir.

**🔹Örneğin;**  
c = 2.[$v_0$, $v_1$] -3.[$v_1$, $v_2$] + 0.[$v_2$, $v_0$] ∈ $C_1$(K)    
zincirine 1-boyutlu zincir(1-chain) denir.

❗️Simpleksler yönlüdür, yani [$v_0$, $v_1$] = -[$v_1$, $v_0$],  
Bu kenarın ters yönde alınması demektir. Dolayısıyla işaret yönle ilgilidir.

Her k için zincir grubu şu şekilde tanımlanır:

$$
C_k(K) = \left\{ \sum a_i [v_{i_0}, \dots, v_{i_k}] \mid a_i \in \mathbb{Z} \right\}
$$

## Sınır Operatörü (Boundary Operator)

Bir simpleksin sınırı onun bir alt boyutundaki yüzlerinden oluşur. Yani bir kenarın sınırı uç noktalardır. Bir üçgenin sınırı kenarlarıdır. Bir tetrahedronun sınırı yüzleri olan üçgenlerdir.

**Tanım:(Sınır operatörü):** Her k-simpleks için onu oluşturan tüm (k-1)-boyutlu yüzleri alıp uygun toplama işlemine sokan operatöre sınır operatörü denir. $δ_k$ notasyonuyla gösterilir.

$$
\partial_k [v_0, v_1, \dots, v_k] = \sum_{i=0}^{k} (-1)^i [v_0, \dots, \hat{v}_i, \dots, v_k]
$$

•&nbsp;$\hat{v}^i$: i. adımda çıkarılan köşe

**Tanım:(Döngü):**  $δ_k$ = 0 ise bu zincire <u>döngü</u> denir.

**🔹Örnek:**  
Bir kenarın sınırı uç noktalarının farkıdır yani; formülü uygulayacak olursak;  
$δ_1$[$v_0$, $v_1$] = [$v_1$] - [$v_0$]

**🔹Örnek:**  
Bir üçgenin sınırı 3 kenarının yönlü toplamına eşittir.     
$δ_2$[$v_0$, $v_1$, $v_2$] = [$v_1$, $v_2$] - [$v_0$, $v_2$] + [$v_0$, $v_1$]   
Dikkat! yönler üçgenin köşe sıralamasına göre değişkenlik gösterebilir.

**🔹Örnek:**  
Bir tetrahedronun sınırı onu çevreleyen 4 adet üçgen yüzeyin yönlü toplamıdır.(Küpün yüzeyi gibi düşün)  
$δ_3$[$v_0$, $v_1$, $v_2$, $v_3$] = [$v_1$, $v_2$, $v_3$] - [$v_0$, $v_2$, $v_3$] + [$v_0$, $v_1$, $v_3$] - [$v_0$, $v_1$, $v_2$] 


**Teorem:(Fundamental Boundary Identity):** Her zincirin sınırının sınırı boştur. Yani $δ_{k-1}$∘$δ_k$ = 0

**Tanım:(Döngü Grubu):** Sınırı sıfır olan k-zincirlerin oluşturduğu gruba k-döngü denir.(cycle)    
$Z_k$ = ker($δ_k$) ile tanımlanır.  


Tanım:(Sınır Grubu): Bir üst boyuttaki zincirlerin sınırlarından oluşan gruba k-sınır denir.(boundary)  
$B_k$ = im($δ_{k+1}$)

→ Sınırı sıfır olan zincirlere döngü dediğimiz için bunları ileride olası delikler olarak adlandıracağız. Fakat bazı döngüler bir üst boyuttaki bir şeyin sınırı olabilir. Bunlara sahte delik denir. Gerçek delik ise sınır olmayan ama sınırı sıfır olan yapıya denir. Gerçek delik ile sahte deliği ayırt edebilmek için yani hangi döngü gerçektir hangi döngü sahtedir sorusuna cevap verebilmek için Homoloji grubu tanımına ihtiyaç vardır.

## Homoloji Grubu

Bazı zincirlerin sınırı sıfırdır. Yani bu zincirlere döngü denir. Fakat bazı döngüler aslında bir şeyin sınırıdır. Yani içi doludur. Bunlara sahte döngü denir. Gerçek döngüler içi dolu olmayandır. Homoloji grubu bu farkı tam olarak cebirsel olarak tanımlar.

**Tanım:(Homoloji):** K simplisel kompleksi üzerinde k'ıncı homoloji grubu:
$$
H_k(K) = Z_k / B_k = \ker(\partial_k) / \operatorname{im}(\partial_{k+1})
$$   
şeklinde tanımlanır. Yani k'ıncı homoloji grubu, k'ıncı döngü grubunun k+1'inci sınır grubuna olan oranıdır.


Yaptığımız şey; sınırı sıfır olan tüm döngüleri($Z_k$) alıp bir üst boyuttaki sınırların oluşturduğu yapıları(B_k) çıkarmak. Böylece elimizde sınırı sıfır olan ama hiçbir şeyin sınırı olmayan döngüler kalır.

**Tanım:(Betti Sayısı):** $H_k$'nın serbest kısmının rütbesine(boyutuna) betti sayısı denir. $β_k$ ile gösterilir.  
$β_k$ = rank($H_k$)

k=0  →  Bağlantı bileşeni sayısı    
k=1  →  Kapalı döngü/delik sayısı   
k=2  →  Kabuk/boşluk sayısı(örn; işi boş küre)

**🔹Örnek:**  
İçi boş bir simplisel kompleks üç kenardan oluşsun. Yani:    
[$v_0$, $v_1$], [$v_1$, $v_2$], [$v_2$, $v_0$]  

• δ(&nbsp;[$v_0$, $v_1$]+[$v_1$, $v_2$]+[$v_2$, $v_0$]&nbsp;)=0 bir döngü olur. Ancak bu döngü, bir üçgenin sınırı değildir. yani ∉ $β_1$     
Bu döngü homolojide sıfır olmayan bir sınıftır. Yani gerçek deliktir. $β_1$ = 1

**🔹Örnek:**  
Bir torus düşünelim. Bu torusun içinde iki temel döngü vardır: Dıştan dönen döngü(etrafı) ve içerden gelen döngü(delik)     
$β_0$ = 1,    
$β_1$ = 2,    
$β_2$ = 1

❗️NOT : Betti sayısı aslında bize şunu söyler:    
•$β_0$ = k için,  Bu yapı k tane ayrı parça içeriyor.   
•$β_1$ = m için, Bu yapı m tane kapalı döngü içeriyor, ama bunlar içi dolu değil.   
•$β_2$ = n için, Bu yapı n tane boşluklu kabuk içeriyor.

## Filtrasyon

Sabit bir simplisel komplekse homoloji uygulamak o yapıdaki delikleri ve kaç tane bağlantı bileşeni, döngü ve yüzey olduğunu gösterir. Peki ya elimizdeki yapı zamanla veya ölçüme bağlı olarak büyüyorsa? Örneğin bir nokta bulutu yavaş yavaş kenarlarla büyüyorsa veya. MRI görüntüsünde piksel değerine göre şekil beliriyorsa? O zaman tek bir homoloji hesaplamak yetersiz kalır. Bu durumda zamanla artan bir simplisel kompleks ailesi kurarız. Bu  yapıya filtrasyon denir.

**Tanım:(Filtrasyon):** Simplisel kompleklerin artan sıralı bir ailesine filtrasyon denir. Φ = $K_0$ ⊆ $K_1$ ⊆ ... ⊆ $K_n$ = $K$   
Burada her $K_i$ bir simplisel komplekstir. Yani kompleks kademeli olarak büyür.

**🔹Örnek:** Diyelim ki elimizde 5 tane nokta var.    
Başlangıçta her nokta ayrı bileşendir yani $β_0$ değeri 5'tir. Aralarındaki mesafeye göre kenar çekmeye başlayalım. Kenarlar oluşunca ayrık bileşenlerimiz birleşiyor. yani $β_0$ değeri azalıyor. Üçgenler kapanınca da döngüler oluşur. Yani $β_1$ değeri artar. Döngüler kapanınca içi dolarsa da $β_1$ değeri azalır. Bu süreç ölçek arttıkça devam eder. Yani zaman yerine burada mesafe eşiği kullanılır. Buna bir filtrasyon denir.

Filtrasyon sayesinde elimizde artık tek bir simplisel kompleks değil, artan bir simplisel kompleks ailesi var.  
$K_0$ ⊆ $K_1$ ⊆ ... ⊆ $K_n$ = $K$   
Her $K_i$ bir simplisel kompleks ve her biri, kendisinden önce gelen kompleksin bir genişletilmesi. Öyleyse bu dizinin her adımında, yani her $K_i$ kompleksinde, bir homoloji hesaplayabiliriz:    
$H_k$($K_0$),   $H_k$($K_1$), ..., $H_k$($K_n$)     
Bunlar her adımda şeklin içindeki k-boyutlu deliklerin sayısını ve yapısını temsil eder. Ancak bu delikler birbirinden bağımsız değildir. Çünkü her $K_i$ kompleksi, $K_{i-1}$'in içinde yer aldığından aralarında doğal cebirsel ilişkiler bulunur.

Filtrasyon ailesi içindeki simplisel komplekslerin oluşturduğu homoloji grupları arasında doğal fonksiyonlar vardır:    

$$
f_i^k : H_k(K_i) \to H_k(K_{i+1})
$$ 

Her fonksiyon, $K_i$ içinde tanımlanmış bir homoloji sınıfını, $K_{i+1}$'deki karşılığına taşır. Bu çok önemlidir çünkü artık sabit bir homoloji grubu değil, zamanla değişen bir yapı incelemeye başlıyoruz.

**Tanım:(Doğum noktası):** Bir delik ilk defa $K_i$'de ortaya çıkarsa, o noktada doğmuştur ve o noktaya doğum noktası denir.  
**Tanım:(Ölüm noktası):** Bir delik daha büyük bir $K_j$'de kapanırsa(örneğin sınır haline gelirse), orada ölmüştür. Buraya ölüm noktası denir.


Bu bakış açısıyla filtrasyon bize "Hangi delik ne zaman oluştu?", "Ne kadar yaşadı?", "Sonunda öldü mü, yoksa filtrasyon boyunca yaşadı mı?" sorularını da cevaplamamıza olanak tanır.

Bundan sonra filtrasyonun her adımında $H_k$($K_i$) grubu içerisindeki delikleri dakip edip, hangi adımda doğup hangi adımda öldüğünü işaretleyip bu bilgiyle bir yaşam süresi profili çıkarabiliriz. Bu profili cebirsel olarak şu yapıyla ifade ederiz:

$$
H_k(K_0) \xrightarrow{f_0^k} H_k(K_1) \xrightarrow{f_1^k} H_k(K_2) \to \cdots \to H_k(K_n)
$$

## Kalıcı Homoloji

Gerçek hayattaki veriler karmaşıktır. Gürültü içerir, farklı ölçeklerde şekil içerir. Biz bu şekillerin en kalıcı en sağlam topolojik özelliklerini bulmak isteriz. Filtrasyon ve homoloji sayesinde gürültüden doğan kısa ömürlü delikleri eleriz. Ayrıca gerçek şekillerin uzun ömürlü topolojik özellikleri yakalanır. Bu analiz sonucunda, şekli sayısal ve görsel olarak özetleyen güçlü araçlar elde edilir.(kalıcılık diyagramı, barkod vs.)

Sabit bir simplisel komplekste homoloji delikleri sayar. Filtrasyon şeklin büyüme sürecini izler. Her farklı bir filtrasyon adımında farklı homoloji grupları ortaya çıkar. Ancak bütün bu adımlardaki delikleri takip etmek için ve hangi deliklerin anlamlı olduğu veya hangi deliklerin gürültülü olduğunu görmek için kalıcı homoloji(persistent homology) tanımını kullanırız.

Kalıcı Homoloji, bir filtrasyon boyunca homolojik özelliklerin(deliklerin) ne zaman ortaya çıktığını ve ne zaman yok olduğunu izler.    
•Delikler kısa süre yaşıyorsa → gürültü  
•Delikler uzun süre yaşıyorsa → anlamlı bir topolojik özellik

Bir $K_i$ ⊆ $K_j$ filtrasyon adımını düşünelim. $K_i$'deki her simpleks aynı zamanda $K_j$'de de bulunduğundan, bu "içerme" (inclusion) ilişkisi, homoloji grupları arasında doğal bir fonksiyon (bir homomorfizma) oluşturur:  

$f_k^{i,j}$: $H_k$($K_i$) → $H_k$($K_j$)

Bu fonksiyon, $K_i$ kompleksindeki p-boyutlu bir deliği (homoloji sınıfını) alır ve onun $K_j$ kompleksindeki karşılığına götürür. Bir delik $K_i$'den $K_j$'ye geçerken ya varlığını sürdürür ya da $K_j$'de içi dolduğu için "sıfır" olur (kaybolur).

İşte bu fonksiyonu kullanarak kalıcı homoloji grubunu tanımlayabiliriz.

Tanım:(k-ıncı Kalıcı Homoloji Grubu): $K$ simplisel kompleksinin bir filtrasyonu verildiğinde i ≤ j için (i,j)-inci k-boyutlu kalıcı homoloji grubu, $H_k^{i,j}$, şu şekilde tanımlanır:   
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$H_k^{i,j}(K)$ = im($f_k^{i,j}$)  
Yani bu grup, i anından j anına kadar kalıcılığını koruyan topolojik özelliklerin bir koleksiyonudur. Bu grubun rankı yani rank($H_k^{i,j}$) bize i anından j anına kadar hayatta kalan k-boyutlu deliklerin sayısını verir. Bu sayıya kalıcı Betti sayısı denir ve $β_p^{i,j}$ ile gösterilir.

**🔹Örnek:**  
Bir γ deliği $K_5$ adımında doğsun. Bu delik $K_6$, $K_7$, $K_8$ adımlarında varlığını sürdürsün. Ancak $K_9$ adımında içi bir üçgenle dolarak ölsün.    
Bu durumda γ deliği $H_1^{5,6}$, $H_1^{5,7}$, $H_1^{5,8}$ gruplarının bir elemanıdır. Bu grupların hepsi γ tarafından üretilen ve rankı 1 olan bir grup içerir. ($β_1^{5,8}$ = 1)    
Ancak $f_1^{5,9}$ fonksiyonu γ'yı sıfıra götürür, çünkü delik $K_9$'da kapanmıştır. Dolayısıyla γ, $H_1^{5,9}$ grubunun bir elemanı değildir. ($β_1^{5,9}$ = 0)

!!! Bir homoloji sınıfı $K_b$'de doğarsa ve $K_d$'de ölürse o zaman bu sınıfın yaşam aralığı; &nbsp;&nbsp;&nbsp;[b,d)&nbsp;&nbsp;&nbsp; şeklinde gösterilir.

**🔹Örnek:**  
Bir filtrasyon tanımlayalım:    
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$K_0$'da 4 ayrı nokta olsun. → $β_0$ = 4  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$K_1$'de noktalar birleşmeye başlasın.    
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$K_2$'de 1. döngü oluşsun. → $β_1$ = 1    
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$K_3$'te döngü hâla yaşasın, kaybolmasın.  
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$K_4$'te üçgen kapansın. Döngümüz sınır olsun.    
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;$K_5$'te döngü sıfır olsun yani ölsün.    

Bu durumda; 

Döngü $H_1$($K_2$)'de doğdu,&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $H_1^{2,3}$, $H_1^{2,4}$ gruplarında yer aldı &nbsp;&nbsp;&nbsp;ve&nbsp;&nbsp;&nbsp; $H_1^{2,5}$'te öldü yani 0 değerini aldı.


Daha genel olarak homoloji grubunu şu şekilde tanımlayabiliriz:     
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; $H_k^{i,j}$ := $Z_i^k$/($B_k^j ∩ Z_k^i$)   
Burada;     
•$Z_i^k$ : $K_i$'deki döngüler yani sınırı sıfır olanlar.   
•$B_k^j$ : $K_j$'deki sınırlar yani daha sonra kapananlar.  

## Kalıcılık Diyagramı

**Tanım:(Kalıcılık Diyagramı):**
$$
\text{Kalıcılık Diyagramı} = \{ (b_i, d_i) \in \mathbb{R}^2 \}
$$
şeklinde tanımlı grafiğe <u>kalıcılık diyagramı</u> denir. Burada;  
•&nbsp;&nbsp;&nbsp;&nbsp;$b_i$ : doğum zamanı   
•&nbsp;&nbsp;&nbsp;&nbsp;$d_i$ : ölüm zamanıdır.    

Doğum-ölüm aralığı ne kadar uzunsa, ona <u>"kalıcı özellik"</u>, <u>"anlamlı özellik"</u> veya <u>"sağlam özellik"</u> adı verilir.

Bir kalıcılık diyagramı $D$, $\mathbb{R}^2$ uzayında sonlu sayıda noktadan oluşan bir çoklu kümedir. Bu kümedeki her noktanın y-koordinatı, x-koordinatından büyüktür $(y>x)$.    
Ayrıca, diyagonal(köşegen) $Δ = \{(x,x) \mid x \in \mathbb{R}\}$ kümesindeki her nokta, sayılabilir sonsuz çoklukta diyagramda yer alır.

Kalıcılık diyagramında x-ekseni doğum zamanını, y-ekseni ölüm zamanını, y=x doğrusu (köşegen) da yaşam süresi 0 olan sınıfları belirtir. Bir sınıf köşegenden ne kadar uzaktaysa gürültü olma ihtimali de o kadar azdır. Yani kalıcı özelliktir.    

**🔹Örnek:**    
Bir kalıcılık diyagramında $(1, 7)$ ve $(2, 2.8)$ noktaları olsun.  
→$(1,7)$ noktası için: Bu sınıf 1. adımda doğdu ve 7. adımda öldü denir. $[1,7)$ ile gösterilir.    
→$(2,2.8)$ noktası için: Bu sınıf 2. adımda doğup çok kısa süre yaşadı denir. Gürültü olduğu kanısına varılır.

❗️Kalıcılık diyagramında tüm noktalar ana köşegenin üstünde yer alır.

Verileri analiz ederken köşegenden uzak noktaları dikkate alırız.

**🔹Örnek:**    
Diyelim ki filtrasyonda şu noktalar geldi:&nbsp;&nbsp;&nbsp;  $(0.5, 2), (1.0, 4.5), (2.3, 2.5), (1.5, 3.5), (0.7, 6.0)$. Sırasıyla bu noktalara $a, b, c, d, e$ diyelim.   
$a$ noktasının ömrü 1.5, kısa ömürlüdür. topolojik özellik olabilir ama sınırdadır.   
$b$ noktasının ömrü 3.5, kalıcıdır önemli bir topolojik özelliktir.     
$c$ noktasının ömrü 0.2, çok kısa ömürlüdür, yüksek ihtimalle gürültüdür.   
$d$ noktasının ömrü 2.0, kısa ömürlü denilemez, önemli olabilir.    
$e$ noktasının ömrü 5.3, çok uzun ömürlüdür. Bu verideki en önemli topolojik özelliktir.

Bu örneği aşağıdaki python koduyla görselleştirelim.

```python
import matplotlib.pyplot as plt
import numpy as np

# Örnekteki (doğum, ölüm) noktaları
points = [(0.5, 2), (1.0, 4.5), (2.3, 2.5), (1.5, 3.5), (0.7, 6.0)]

# Veriyi çizim için hazırlayalım (x ve y eksenleri için ayırma)
# x ekseni: doğum zamanları
births = [p[0] for p in points]
# y ekseni: ölüm zamanları
deaths = [p[1] for p in points]

# Çizim alanını oluşturalım
fig, ax = plt.subplots(figsize=(8, 8))

# 1. Diyagonal (y=x) çizgisini çizelim
# En yüksek ölüm değerine göre bir limit belirleyelim
limit = max(deaths) + 1
diag_line = np.linspace(0, limit, 100)
ax.plot(diag_line, diag_line, color='gray', linestyle='--', linewidth=2, label='$y=x$ (Gürültü sınırı)')

# 2. Kalıcılık noktalarını (doğum, ölüm) grafiğe ekleyelim
ax.scatter(births, deaths, s=100, c='royalblue', zorder=5, label='Topolojik Özellikler')

# 3. Grafik ayarlarını ve etiketleri ekleyelim
ax.set_title("Örnek Kalıcılık Diyagramı", fontsize=16)
ax.set_xlabel("Doğum (Birth)", fontsize=12)
ax.set_ylabel("Ölüm (Death)", fontsize=12)

# Eksenlerin başlangıç noktasını (0,0) yapalım ve grid ekleyelim
ax.set_xlim(left=0)
ax.set_ylim(bottom=0)
ax.grid(True, linestyle='--', alpha=0.6)

# Grafiğin en/boy oranını eşitleyerek y=x çizgisinin 45 derece görünmesini sağlayalım
ax.set_aspect('equal', adjustable='box')

# Hangi noktanın hangisi olduğunu görmek için etiket ekleyelim (isteğe bağlı)
for p in points:
    ax.text(p[0] + 0.05, p[1] + 0.05, f'({p[0]}, {p[1]})', fontsize=9)

# Açıklama kutusunu (legend) gösterelim
ax.legend()

# Grafiği göster
plt.show()
```

**🔹Örnek**

Bir teknoloji şirketi, "Synth-5" adında yeni bir akıllı telefon modelini piyasaya sürdü. Piyasaya sürülmesinden sonraki ilk 24 saat içinde atılan binlerce tweet'i topladığımızı ve bu tweet'lerin içeriğini analiz etmek istediğimizi düşünelim. Amacımız, insanların bu yeni telefon hakkında ne konuştuğunu, ana tartışma konularının ne olduğunu ve bu konuların birbiriyle nasıl ilişkili olduğunu anlamak.    
Her bir tweet'i, gelişmiş bir NLP (Doğal Dil İşleme) modeli kullanarak 128 boyutlu bir vektöre dönüştürdük. Bu uzayda, anlam olarak birbirine yakın olan tweet'ler, birbirine yakın vektörler olarak konumlanır.

Aşağıda, bu yüksek boyutlu verinin 2 boyuta indirgenmiş halini temsil eden, 35 tweet'ten oluşan simüle edilmiş bir veri seti bulunmaktadır.

```python
import numpy as np
import matplotlib.pyplot as plt

# Rastgeleliği sabitleyelim ki her çalıştırmada aynı sonuç çıksın
np.random.seed(42)

# 35 veri noktası oluşturalım
# Küme 1: "Kamera Harika" diyenler (10 nokta)
kamera_grubu = np.random.randn(10, 2) * 0.5 + np.array([2, 8])

# Küme 2: "Batarya Ömrü Kötü" diyenler (12 nokta)
batarya_grubu = np.random.randn(12, 2) * 0.6 + np.array([8, 2])

# Küme 3: "Tasarım ve Ekran" üzerine konuşanlar (8 nokta)
tasarim_grubu = np.random.randn(8, 2) * 0.4 + np.array([8, 8])

# Gürültü/Alakasız Tweetler (5 nokta)
gurultu = np.random.rand(5, 2) * 12

# Tüm noktaları birleştirelim
veri_seti = np.vstack([kamera_grubu, batarya_grubu, tasarim_grubu, gurultu])

# Veri setini görselleştirelim
plt.figure(figsize=(10, 8))
plt.scatter(veri_seti[:, 0], veri_seti[:, 1], s=50, alpha=0.8)
plt.title("Tweet Veri Setinin 2D Temsili", fontsize=16)
plt.xlabel("Vektör Boyut 1")
plt.ylabel("Vektör Boyut 2")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

Yukarıda 3 ana küme ve etrafa dağılmış birkaç nokta bulutu görüyoruz. Bu nokta bulutuna filtrasyon uyguladığımızı hayal edelim.

• Başlangıçta her noktamız ayrı bileşendir. Yani $β_0$ = 35.     
• Noktalar arasında mesafeyi ölçen bir ε yarıçapını yavaş yavaş artırıyoruz.    
• İki nokta arasındaki mesafe ε'dan küçükse, aralarına bir kenar (1-simpleks) ekliyoruz. Bu, $β_0$ sayısını azaltır.    
• Üç nokta birbirine kenarlarla bağlıysa, aralarına bir üçgen (2-simpleks) ekliyoruz. Bu, var olan döngüyü öldürebilir.

```python
import numpy as np
import matplotlib.pyplot as plt
from gtda.homology import VietorisRipsPersistence

# Veri ve Homoloji Hesaplama
np.random.seed(42)
kamera_grubu = np.random.randn(10, 2) * 0.5 + np.array([2, 8])
batarya_grubu = np.random.randn(12, 2) * 0.6 + np.array([8, 2])
tasarim_grubu = np.random.randn(8, 2) * 0.4 + np.array([8, 8])
gurultu = np.random.rand(5, 2) * 12
veri_seti = np.vstack([kamera_grubu, batarya_grubu, tasarim_grubu, gurultu])
persistence = VietorisRipsPersistence(homology_dimensions=(0, 1))
diagrams = persistence.fit_transform(veri_seti[np.newaxis, :, :])
diagram = diagrams[0]

h0_pts = diagram[diagram[:, 2] == 0]
h1_pts = diagram[diagram[:, 2] == 1]


h0_pts_jittered = h0_pts.copy()
jitter_strength = 0.05
h0_pts_jittered[:, 0] += np.random.rand(len(h0_pts)) * jitter_strength

fig, ax = plt.subplots(figsize=(8, 8))

# Noktaları çizdirelim
ax.scatter(h0_pts_jittered[:, 0], h0_pts_jittered[:, 1], label='H0 (Bileşenler)', c='tomato', s=50, alpha=0.7)
ax.scatter(h1_pts[:, 0], h1_pts[:, 1], label='H1 (Döngüler)', c='turquoise', s=50, alpha=0.7)

# Diyagonal çizgiyi çizelim
max_val = np.max(diagram) * 1.05
ax.plot([0, max_val], [0, max_val], 'k--', label='Köşegen Doğrusu (y=x)', linewidth=2)

# Grafik ayarları
ax.set_xlabel("Doğum (Birth)", fontsize=12)
ax.set_ylabel("Ölüm (Death)", fontsize=12)
ax.set_title("Kalıcılık Diyagramı", fontsize=16)
# X ekseninin başlangıcını hafifçe negatiften başlatarak 0 çizgisini daha belirgin yapalım
ax.set_xlim(-0.1, max_val) 
ax.set_ylim(0, max_val)
ax.set_aspect('equal', adjustable='box')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

Yukarıdaki diyagramda iki farklı renk var:

• Kırmızı Noktalar: 0-boyutlu homolojiyi ($H_0$), yani bağlantı bileşenlerini temsil eder.      
• Turkuaz Noktalar: 1-boyutlu homolojiyi ($H_1$), yani döngüleri veya delikleri temsil eder.

**Kırmızı noktalar**       

→Diyagramın sol altında $y=x$ doğrusuna yapışık bir sürü kırmızı nokta var. Bunlar tek tek tweet'lerin veya çok küçük grupların, filtrasyonun ilk adımında hemen birleşerek ölmesinden kaynaklanır. Bunlar, analizimiz için anlamlı bir bilgi taşımaz, sadece verinin başlangıçtaki dağınıklığını gösterir.

→Grafiğin üst kısmında köşegenden uzakta duran birkaç tane kırmızı nokta görüyoruz. Bu noktaların ölüm zamanı diğerlerine göre çok yüksektir, yani yaşan süreleri çok uzundur. Daha da kısıtlayacak olursak, en üstteki 3 nokta, _Kamera_, _Batarya_ ve _Tasarım_ gruplarının kendilerini temsil eder. Bu kümeler o kadar sağlam ve kalıcıdır ki, diğer tüm küçük birleşmeler bittikten sonra bile vardıklarını sürdürürler fakat eninde sonunda ölürler.

**Turkuaz noktalar**

→Diyagramda köşegene yakın bir turkuaz nokta kümesi var. Bunlar filtrasyon esnasında rastgele oluşan ve ardından hemen kapanan anlamsız küçük deliklerdir. Bunlar ihmal edilir.

→Diyagramın sağ üstünde köşegenden biraz uzak bir tane turkuaz nokta var. Bu döngünün ilk oluştuğu an(4.6) _Kamera_, _Batarya_ ve _Tasarım_ kümelerinin ε yarıçapı 4.6 değerine ulaştığında birbirlerine dokunarak ortada boşluk olacak bir şekilde bir 'halka' oluşturduğu anlamına gelir.
Döngünün kapandığı an(5.2) döngüyü oluşturan simplekslerin arasına daha yüksek boyutlu bir simpleksin girmesiyle olur. Senaryomuzda bu, ε yarıçapı 5.2'ye ulaştığında, üç ana kümeden birer noktanın birbirine yeterince yaklaşıp aradaki boşluğu bir "üçgenle doldurduğu" anlamına gelir.

**Özet olarak;**

Bir $H_1$ noktasının koordinatları, döngünün hangi ε değeri için başlayıp bittiğini söyler. Köşegene uzaklığı o döngünün ne kadar önemli ve yapısal olduğu hakkında yorum yapmamızı sağlar.

## Barkod Diyagramı


**Tanım:(Barkod Diyagramı):** Kalıcı homoloji sınıflarını [doğum, ölüm) aralıkları şeklinde yatay çizgilerlerle  gösteren görsel bir temsile barkod diyagramı(çubuk diyagramı) denir. Her çubuk, bir homoloji sınıfını(yani delikleri) temsil eder.     
Her bir topolojik özellik, doğum anında başlayan ve ölüm anında biten bir yatay çubuk (barkod) olarak gösterilir.

Bir önceki örnekte yapılanların aynısını yapalım, bu sefer diyagramını barkod diyagramı olarak çizelim;

```python
import numpy as np
import matplotlib.pyplot as plt
from gtda.homology import VietorisRipsPersistence
from gtda.plotting import plot_diagram

# Kendi plot_barcode fonksiyonumuzu tanımlayalım
def custom_plot_barcode(persistence_diagram, homology_dim):
    """
    Verilen bir homoloji boyutu için kalıcılık barkodunu çizer.
    """
    # Sadece istenen boyuttaki özellikleri filtrele
    bars = persistence_diagram[persistence_diagram[:, 2] == homology_dim]
    
    # Sonsuz ölüm değerlerini grafikte daha anlamlı bir üst sınırla değiştir
    finite_death = bars[:, 1][np.isfinite(bars[:, 1])]
    inf_death_replace = np.max(finite_death) * 1.1 if len(finite_death) > 0 else 1
    bars[np.isinf(bars[:, 1]), 1] = inf_death_replace
    
    plt.figure(figsize=(10, 4))
    for i, (birth, death, _) in enumerate(bars):
        # Bu satırda renk parametresi eklendi
        plt.plot([birth, death], [i, i], lw=2, color='k') 
    
    plt.title(f'H{homology_dim} için Kalıcılık Barkodu')
    plt.xlabel('Filtrasyon değeri (Radius)')
    plt.yticks([]) # y eksenindeki tekil bar etiketlerini gizle
    plt.show()

# --- Veri oluşturma kısmı ---
np.random.seed(42)
kamera_grubu = np.random.randn(10, 2) * 0.5 + np.array([2, 8])
batarya_grubu = np.random.randn(12, 2) * 0.6 + np.array([8, 2])
tasarim_grubu = np.random.randn(8, 2) * 0.4 + np.array([8, 8])
gurultu = np.random.rand(5, 2) * 12
veri_seti = np.vstack([kamera_grubu, batarya_grubu, tasarim_grubu, gurultu])

# --- Homoloji hesaplama ve çizdirme ---
persistence = VietorisRipsPersistence(homology_dimensions=(0, 1))
diagrams = persistence.fit_transform(veri_seti[np.newaxis, :, :])

diagram = diagrams[0]

# H0 ve H1 için siyah renkli barkodları çizdir
custom_plot_barcode(diagram, homology_dim=0)
custom_plot_barcode(diagram, homology_dim=1)
```

Yukarıda görüldüğü üzere, bir önceki örnekteki yorumların aynısını farklı bir bakış açısıyla yapabiliriz.

## Betti Eğrileri

Kalıcılık diyagramları ve barkodlar, verinin topolojik özetini bize sunar. Ancak bu özet, "değişken sayıda noktadan veya aralıktan" oluşur. Bu, makine öğrenmesi için bir zorluktur. Çünkü makine öğrenmesi algoritmaları genellikle sabit boyutlu _vektörleri_ girdi olarak ister. Bu yüzden Betti eğrilerini kullanırız.

**Tanım:(Betti Eğrisi):** Belirli bir homoloji boyutu (örneğin $H_0$, $H_1$) için, filtrasyonun herhangi $t$ anında hayatta olan özelliklerin sayısını gösteren fonksiyona <u>betti eğrisi</u> denir.
$$
    β_k(t) = \text {Filtrasyonun  t anında $H_k$ grubundaki sınıf sayısı}
$$

Fonksiyonda bir özellik $b$ anında doğduğunda, eğrinin değeri 1 artar. Bir özellik $d$ anında öldüğünde, eğrinin değeri 1 azalır. Sonuç olarak betti eğrisi aslında bize topolojik özellik sayısının nasıl değiştiğini gösteren bir <u>adım fonksiyonudur(step function)</u>.

Eğrideki pik noktası yani en zirvedeki yer, filtrasyonun o "zamanında"(yani o ε ölçeğinde) aynı anda çok sayıda topolojik özelliğin hayatta olduğunu gösterir. Bu, verinin o ölçekte yapısal olarak en karmaşık olduğu an olabilir.     
Eğride uzun bir düzlük varsa belirli sayıdaki özelliğin stabil bir şekilde varlığını sürdürdüğünü gösterir. Bu, genellikle verinin en önemli ve en kalıcı yapılarının bir işaretidir.       
Hızla azalan bir eğri varsa özelliklerinin kalıcılığının olmadığını söyleriz.

🔔 Betti eğrisini kullanmanın en önemli avantajı, bir fonksiyonu kolayca sabit boyutlu bir vektöre dönüştürebilmemizdir. Mesela, betti eğrisini (0,10) aralığında 100 eşit noktada örneklediğimizi varsayalım. O halde elimizde bu veri setinin topolojisini temsil eden 100 boyutlu bir vektör olur. Dolayısıyla bu vektörleri makine öğrenmesi algoritmalarında kullanabiliriz.

Yine aynı tweet veri seti için betti eğrisi çizelim;

```python
import numpy as np
import matplotlib.pyplot as plt
from gtda.homology import VietorisRipsPersistence
from gtda.diagrams import BettiCurve

# --- VERİ OLUŞTURMA VE HOMOLOJİ HESAPLAMA ---
np.random.seed(42)
kamera_grubu = np.random.randn(10, 2) * 0.5 + np.array([2, 8])
batarya_grubu = np.random.randn(12, 2) * 0.6 + np.array([8, 2])
tasarim_grubu = np.random.randn(8, 2) * 0.4 + np.array([8, 8])
gurultu = np.random.rand(5, 2) * 12
veri_seti = np.vstack([kamera_grubu, batarya_grubu, tasarim_grubu, gurultu])

persistence = VietorisRipsPersistence(homology_dimensions=(0, 1))
diagrams = persistence.fit_transform(veri_seti[np.newaxis, :, :])

# --- BETTI EĞRİSİ HESAPLAMA ---
BC = BettiCurve()
X_betti_curves = BC.fit_transform(diagrams)

# X ve Y ekseni verilerini ayrı değişkenlere alalım
x_axis_data = BC._samplings[0]
y_axis_h0 = X_betti_curves[0, 0, :]
y_axis_h1 = X_betti_curves[0, 1, :]

#reshape(-1) ile verileri 1-boyutlu hale getirelim <<<
# Bu fonksiyon, şekli ne olursa olsun veriyi dümdüz bir listeye çevirir.
x_flat = x_axis_data.reshape(-1)
y_h0_flat = y_axis_h0.reshape(-1)
y_h1_flat = y_axis_h1.reshape(-1)

# Şimdi çizim yapalım
plt.figure(figsize=(10, 6))

plt.plot(x_flat, y_h0_flat, label='$H_0$ Betti Eğrisi (Bileşenler)', color='tomato', linewidth=2)
plt.plot(x_flat, y_h1_flat, label='$H_1$ Betti Eğrisi (Döngüler)', color='turquoise', linewidth=2)

# Grafik Ayarları
plt.title('Betti Eğrileri', fontsize=16)
plt.xlabel('Filtrasyon Parametresi ($\epsilon$)', fontsize=12)
plt.ylabel('Hayattaki Özellik Sayısı (Betti Sayısı)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

Diğer 2 küme de kendi içlerinde veya birbirine yakın bir şekilde iki ayrı küçük ama stabil konu grubu olabilir. Mesela telefonun fiyatı veya boyutu. $y=3$ olduğunda ise bu 2 konunun kaybolduğunu ve geriye sağlam 3 ana konunun kaldığını gösteriyor.

## Stabilite Teoremi

Her şeyi yaptık fakat eninde sonunda şu soruyu kendimize sorarız: _"Verim biraz gürültülüyse, eğer ölçümümde küçük bir hata varsa, diyagramlar değişip beni yanıltabilir mi?"_      
Stabilite Teoremi der ki:       
Giriş verisindeki küçük bir değişiklik, çıktı olan kalıcılık diyagramında sadece küçük bir değişikliğe yol açar.

Daha matematiksel bir tanım yapmamız için birkaç mesafe tanımı yapmamız lazımdır.

**Tanım:(Hausdorff Mesafesi):** $(Z, d_Z)$ bir metrik uzay ve $X, Y$ bu uzayın boş olmayan iki kapalı alt kümesi olsun. $X ve Y$ arasındaki **Hausdorff mesafesi** şu şekilde tanımlanır:
$$
d_H(X, Y) = \max \left\{ \sup_{x \in X} d(x, Y), \sup_{y \in Y} d(y, X) \right\}
$$

Burada $d(p,S)$ = $\inf_{s \in S} d_Z(p, s)$, bir $p$ noktasının bir S kümesine olan en kısa mesafesidir.

**🔹 Örnek :** $X$ ve $Y$ iki nokta kümesi olsun. Aralarındaki Hausdorff mesafesi $d_H(X,Y)$ iki yönlü bir ölçümdür, yani:


&nbsp;&nbsp;&nbsp;1-) $X$'teki her bir $x$ noktası için, $Y$'deki en yakın $y$ noktasının en büyük olanı    
&nbsp;&nbsp;&nbsp;2-) $Y$'deki her bir $y$ noktası için, $X$'deki en yakın $x$ noktasının en büyük olanı        
Bu iki durumun en büyüğü Hausdorff mesafesidir.


**Tanım(Gromov-Hausdorff Mesafesi)**: $(X, d_X)$ ve $(Y, d_Y)$ boş olmayan iki kompakt metrik uzay olsun. X ve Y arasındaki **Gromov-Hausdorff mesafesi** şu şekilde tanımlanır:
$$
d_{GH}(X, Y) = \inf \{ d_H^Z(f(X), g(Y)) \}
$$

Burada infimum, tüm metrik uzaylar $(Z, d_Z)$ ve $X$ ile $Y$'yi $Z$ uzayına gömen tüm izometrik gömme fonksiyonları $f : X → Z$ &nbsp;&nbsp;ve $g : Y → Z$ üzerinden alınır.

**🔹 Örnek :** İki farklı adada olduğumuzu hayal edelim. Bir adadaki herhangi bir noktadan diğer adadaki en yakın noktaya yüzmek ne kadar sürer? Ve tam tersi, diğer adadan bu adaya en yakın noktaya yüzmek ne kadar sürer? Bu "en uzun süren en kısa yüzme mesafesi" bize bu iki adanın birbirine ne kadar benzediğini söyler. Eğer adalar iç içe geçmişse bu mesafe sıfıra yakındır. Eğer çok uzaksalar mesafe büyüktür. Yani Gromov-Hausdorff mesafesi tam olarak iki nokta bulutunun (veya daha genel olarak metrik uzayın) birbirine ne kadar "benzemediğini" ölçer. Bunu, iki kümeyi de aynı metrik uzaya koyup aralarındaki Hausdorff mesafesine bakarak yapar ve en iyi sonucu verecek şekilde optimize eder.

**Tanım:(Bottleneck Mesafesi):** $D_1$ ve $D_2$ iki kalıcılık diyagramı olsun. $D_1$ ve $D_2$ arasındaki bir **kısmi eşleşme (partial matching)** $η$, $D_1$'in bir alt kümesinden $D_2$'nin bir alt kümesine giden bir bijeksiyondur. $D_1$ ve $D_2$ arasındaki **Bottleneck mesafesi**, şu şekilde tanımlanır:
$$
d_B(D_1, D_2) = \inf_{\eta} \max \left\{ \sup_{p \in \text{dom}(\eta)} \|p - \eta(p)\|_{\infty}, \sup_{p \in D_1 \setminus \text{dom}(\eta)} \frac{\|p - \pi(p)\|_{\infty}}{\sqrt{2}}, \sup_{q \in D_2 \setminus \text{im}(\eta)} \frac{\|q - \pi(q)\|_{\infty}}{\sqrt{2}} \right\}
$$
❗️Genellikle diyagonal mesafesi $\frac{d(p, π(p))}{\sqrt{2}}$ yerine basitleştirilmiş hali olan $d(p, proj(p))$ kullanılır, ki bu $\frac{y-x}{2}$'ye denktir.

Sezgisel Fikir: Elimizde iki farklı kalıcılık diyagramı olsun. Bu iki diyagramı nasıl karşılaştırırız? Fikir, bu diyagramdaki bir noktaları diğer diyagramdaki noktalara aşağıdaki koşullar sağlanmak üzere eşleştirmektir.     
&nbsp;&nbsp;&nbsp;&nbsp;• Her nokta en fazla bir noktayla eşleşebilir.      
&nbsp;&nbsp;&nbsp;&nbsp;• Eşleşmeyen noktalar, diyagonal çizgisindeki en yakın izdüşümüyle eşleştirilir.    
&nbsp;&nbsp;&nbsp;&nbsp;• Her eşleşmenin bir maliyeti vardır. Bu maliyet, eşleşen iki nokta arasınaki en büyük mesafedir.(genellikle $L_{\infty}$-normu yani $max(|x_1-x_2|, |y_1-y_2|)$ kullanılır)       

Bottleneck mesafesi, bu eşleştirmeyi en büyük maliyeti en aza indirecek şekilde yapan mükemmel eşleşmeyi arar. Bu en iyi eşleşmedeki maliyet, iki diyagram arasındaki Bottleneck mesafesidir.

Şimdi stabilite teoremini verebiliriz.

**Teorem:(Stabilite Teoremi):** İki veri kümesinin kalıcılık diyagramları arasındaki Bottleneck Mesafesi, bu iki veri setinin kendileri arasındaki Gromov-Hausdorff Mesafesinden hiçbir zaman daha büyük olamaz.
$$
d_B(Dgm(X), Dgm(Y)) \le d_{GH}(X, Y)
$$

Yani, diyagramlar arasındaki fark, veriler arasındaki farktan daha büyük olamaz. Veriyi ne kadar bozarsan, diyagram da o kadar bozulur, daha fazla değil.

**💥Sonuç:**        
1. <u>Gürültüye Karşı Dayanıklılık:(Robustness):</u>   Mesela görüntümüzde hafif bir "parazit" veya ölçüm gürültüsü varsa, bu gürültü kalıcılık diyagramını mahvetmez. Sadece diyagonalin çok yakınına birkaç tane anlamsız kısa ömürlü nokta ekler. Diyagonalden uzakta duran ana sinyal olarak tespit edilen noktalar ise yerinden oynamaz. Böylece gürültünün içindeki gerçek yapı bozulmamış olur.
2. <u>Karşılaştırılabilirlik:(Comparability):</u>   İki farklı hastanın BT görüntüsünü analiz ettiğimizi düşünelim, eğer bu iki hastanın da BT'deki lezyonların şekli birbirine yakınsa, stabilite teoremi sayesinde onların kalıcılık diyagramlarının da birbirine çok benzeyeceğini biliriz. Çünkü aralarındaki Bottleneck mesafesi küçük olacaktır. Bu, farklı verileri topolojik olarak karşılaştırmamızı ve "şekil benzerliğine göre" gruplamamızı sağlar.
